## 💾 The data: sleep_health_data.csv

SleepInc has provided you with an anonymized dataset of sleep and lifestyle metrics for 374 individuals. This dataset contains average values for each person calculated over the past six months. The data is saved as `sleep_health_data.csv`.

The dataset includes 13 columns covering sleep duration, quality, disorders, exercise, stress, diet, demographics, and other factors related to sleep health. 

| Column | Description |
|---------|----------------------------------------|  
| `Person ID` | An identifier for each individual. |
| `Gender` | The gender of the person (Male/Female). |  
| `Age` | The age of the person in years. |
| `Occupation` | The occupation or profession of the person. |
| `Sleep Duration (hours)` | The average number of hours the person sleeps per day. |
| `Quality of Sleep (scale: 1-10)` | A subjective rating of the quality of sleep, ranging from 1 to 10. |
| `Physical Activity Level (minutes/day)` | The average number of minutes the person engages in physical activity daily. |  
| `Stress Level (scale: 1-10)` | A subjective rating of the stress level experienced by the person, ranging from 1 to 10. |
| `BMI Category` | The BMI category of the person (e.g., Underweight, Normal, Overweight). |
| `Blood Pressure (systolic/diastolic)` | The average blood pressure measurement of the person, indicated as systolic pressure over diastolic pressure. |
| `Heart Rate (bpm)` | The average resting heart rate of the person in beats per minute. |
| `Daily Steps` | The average number of steps the person takes per day. |
| `Sleep Disorder` | The presence or absence of a sleep disorder in the person (None, Insomnia, Sleep Apnea). |

### Find occupation with the lowest sleep duration

Set to lowest_sleep_occ

```python
df_job = df.groupby("Occupation").mean("Sleep Duration")
df_job.head()
```

```python
df_job.index
```

```python
lowest_sleep_occ = df_job[["Sleep Duration"]].sort_values(by="Sleep Duration", ascending=True).index[0]
lowest_sleep_occ
```

```python
df_job.loc['Sales Representative', 'Sleep Duration']
```

### Find occupation with the lowest sleep quality

Set to lowest_sleep_quality_occ

```python
df_sq_job = df.groupby("Occupation").mean("Quality of Sleep")
df_sq_job.head()
```

```python
lowest_sleep_quality_occ = df_sq_job[["Quality of Sleep"]].sort_values(by="Quality of Sleep", ascending=True).index[0]
lowest_sleep_quality_occ
```

```python
df_sq_job.loc['Sales Representative', 'Quality of Sleep']
```

```python
same_occ = lowest_sleep_occ == lowest_sleep_quality_occ
same_occ
```

### Explore BMI Category can affect sleep disorder rates, find the ratio of app users in each BMI Category

Store in dictionary named bmi_insomnia_ratios

```python
df["Sleep Disorder"].value_counts()
```

```python
df_bmi = df.groupby("BMI Category").count()
df_bmi
```

```python
bmi_count_series = df["BMI Category"].value_counts()
bmi_count_series
```

```python
mask = (df["BMI Category"] == "Normal") & (df["Sleep Disorder"] == "Insomnia")
normal_count = mask.sum()
normal_count
```

```python
bmi_insomnia_ratios = {}
bmi_insomnia_ratios["Normal"] = normal_count / bmi_count_series["Normal"]
bmi_insomnia_ratios["Normal"] 
```

```python
mask2 = (df["BMI Category"] == "Obese") & (df["Sleep Disorder"] == "Insomnia")
obese_count = mask2.sum()
obese_count
```

```python
bmi_insomnia_ratios["Obese"] = obese_count / bmi_count_series["Obese"]
```

```python
mask3 = (df["BMI Category"] == "Overweight") & (df["Sleep Disorder"] == "Insomnia")
overweight_count = mask3.sum()
overweight_count
```

```python
bmi_insomnia_ratios["Overweight"] = overweight_count / bmi_count_series["Overweight"]
```

```python
bmi_insomnia_ratios
```

## REVISITING THE PROJECT 

```python
# groupby("Occupation) -> groups by occupation
# mean() -> computes average sleep duration per occupation
# idxmin() -> returns the occupation with the smallest mean value
lowest_sleep_occ = df.groupby("Occupation")["Sleep Duration"].mean().idxmin()
lowest_sleep_occ
```

```python
lowest_sleep_quality_occ = df.groupby("Occupation")["Quality of Sleep"].mean().idxmin()
lowest_sleep_quality_occ
```

```python
same_occ = lowest_sleep_occ == lowest_sleep_quality_occ
same_occ
```

```python
# Count total people per BMI category
bmi_count_series = df["BMI Category"].value_counts()
bmi_count_series
```

```python
# Count insomnia cases per BMI category
insomnia_count_series = df[df["Sleep Disorder"] == "Insomnia"]["BMI Category"].value_counts()
insomnia_count_series
```

```python
# Compute ratios
bmi_insomnia_ratios = (insomnia_count_series / bmi_count_series).round(2)
bmi_insomnia_ratios
```

```python
# Convert to dictionary
bmi_insomnia_ratios = bmi_insomnia_ratios.to_dict()
bmi_insomnia_ratios
```
