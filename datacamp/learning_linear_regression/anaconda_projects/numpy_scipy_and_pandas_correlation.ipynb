```python
import numpy as np
```

```python
x = np.arange(10,20)
x
```

```python
y = np.array([2, 1, 4, 5, 8, 12, 18, 25, 96, 48])
y
```

```python
r = np.corrcoef(x, y)
r
```

```python
r[0,1]
```

```python
r[1,0]
```

```python
import scipy.stats
```

```python
scipy.stats.pearsonr(x,y)
```

```python
scipy.stats.spearmanr(x,y) # Spearman's rho
```

```python
scipy.stats.kendalltau(x,y) # Kendall's tau
```

```python
scipy.stats.pearsonr(x,y)[0]
```

```python
scipy.stats.spearmanr(x,y)[0]
```

```python
scipy.stats.kendalltau(x,y)[0]
```

```python
# use dot notation
scipy.stats.spearmanr(x, y).correlation
```

```python
scipy.stats.kendalltau(x, y).correlation
```

```python
# unpacking the return value
r, p = scipy.stats.pearsonr(x, y)
r
```

```python
import pandas as pd
```

```python
x = pd.Series(range(10,20))
x
```

```python
y = pd.Series([ 2,  1,  4,  5,  8, 12, 18, 25, 96, 48])
y
```

```python
x.corr(y)
```

```python
y.corr(x)
```

```python
x.corr(y, method='spearman')
```

```python
x.corr(y, method='kendall')
```

```python
# do linear regression
result = scipy.stats.linregress(x,y)
print(f"result.splope: {result.slope}")
```

```python
print(f"result.intercept: {result.intercept}")
print(f"result.rvalue: {result.rvalue}")
print(f"result.pvalue: {result.pvalue}")
print(f"result.stderr: {result.stderr}")
```
