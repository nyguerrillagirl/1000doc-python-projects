### Example 12.5 
A random sample of 11 statistics students produced the data in Table 12.1, where x is the third exam score out of 80 and y is the final exam score out of 200. Can you predict the final exam score of a random student if you know the third exam score?

<img src="table_12_1.png" width="25%">

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress, pearsonr
```

```python
# Data
# x - third exam score
x = np.array([65, 67, 71, 71, 66, 75, 67, 70, 71, 69, 69])
# y - final exam score
y = np.array([175, 133, 185, 163, 126, 198, 153, 163, 159, 151, 159])
```

```python
# Linear regression
slope, intercept, r_value, p_value, std_err = linregress(x, y)
```

```python
# Predicted values
y_pred = slope * x + intercept

# Pearson correlation
r, p = pearsonr(x, y)
```

```python
# Plot
plt.figure(figsize=(8, 6))

plt.scatter(x, y, color='blue', label='Data')

# Regression line
x_line = np.linspace(x.min(), x.max(), 100)
y_line = slope * x_line + intercept

plt.plot(
    x_line,
    y_line,
    color='red',
    linewidth=2,
    label=f'y = {slope:.2f}x + {intercept:.2f}'
)

plt.xlabel('Third Exam Score')
plt.ylabel('Final Exam Score')
plt.title('Third Exam vs Final Exam')
plt.legend()
plt.grid(True)

plt.show()
```

```python
print(f"Slope = {slope:.4f}")
print(f"Intercept = {intercept:.4f}")
print(f"Correlation r = {r:.4f}")
print(f"R-squared = {r**2:.4f}")
print(f"p-value = {p:.6f}")
```
